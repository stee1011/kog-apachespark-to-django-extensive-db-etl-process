```python
df = spark.read.csv("Comprehensive_Banking_Database.csv", header=True,inferSchema=True)
```

```python
df.orderBy(col("Age").desc()).select("Age", "Last Name").limit(5).show()
```

```python
df.describe(["Age", "Transaction Amount"]).show()
```

```python
df = df.drop("Feedback ID", "Feedback Date", "Feedback Type")
```

```python
df.select("Loan Status").show(5)
```

```python
df.filter(col("Loan Status") == "Approved").select("Loan Type").show(10)
```

```python
df.groupBy(col("Loan Type")) \
    .agg(
        avg(col("Loan Amount")) \
            .alias("loan-avg")
    ) \
    .orderBy(col("loan-avg").desc()) \
    .select("Loan Type", "loan-avg") \
    .cache() \
    .limit(7) \
    .show()
```

```python
data = df.select("First Name","Age","Gender","City","Account Type","Account Balance","Transaction Type","Transaction Amount","Account Balance After Transaction","Loan Amount","Loan Type","Rewards Points","Anomaly")
```

```python
data.limit(5).show()
```

```python
data.printSchema()
```

```python
data.filter((col("Transaction Type") == "Withdrawal") & (col("Account Balance") - col("Transaction Amount") < col("Account Balance After Transaction"))).show()
```

```python
from pyspark.sql.functions import col, sum as _sum

nulls = bigdata.select([
    _sum(col(c).isNull().cast("int")).alias(c)
    for c in bigdata.columns
])
nulls.show(truncate=False)
```

```python
bigdata = bigdata.repartition(8)
```
